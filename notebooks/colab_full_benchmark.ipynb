```python
# Run adversarial benchmark with Model Poisoning attack and Trimmed Mean defense
import yaml
with open('configs/smoke.yaml') as f:
    cfg = yaml.safe_load(f)

cfg['experiment_id'] = 'colab_adversarial_demo'
cfg['attack'] = {'type': 'model_poison', 'malicious_fraction': 0.2, 'strength': -1.0}
cfg['defense'] = {'type': 'trimmed_mean', 'trim_ratio': 0.2}
cfg['output_dir'] = 'results/colab_adversarial_demo'

with open('configs/colab_adv.yaml', 'w') as f:
    yaml.dump(cfg, f)

!fedids-bench run --config configs/colab_adv.yaml
```
